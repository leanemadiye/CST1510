## Topic 1 · Case and whitespace
*Walkthrough: `01 - Case and Whitespace`*

### D1. Case changes comparisons

1. Create `a = "SRV-01"` and `b = "srv-01"`.
2. Print `a == b`.
3. Print the same comparison again, with `.lower()` on both sides.

**Expected output:** `False`, then `True`

````python
#D1
"""python comparisons are case sensetive so "SRV-01" != "srv-01"""

a = "SRV-01"
b = "srv-01"
print(a == b)   #false because a is in uppercase and b is lowercase
print(a.lower() == b.lower())   
````

### D2. `.upper()`, `.lower()`, `.title()`

1. Create `name = "record check console"` and print it with `.title()`, then with `.upper()`.
2. Create `original = "SRV-01"`. Print `original.lower()`, then print `original` again.
3. Add a comment explaining why `original` did not change.

**Expected output:**
```
Record Check Console
RECORD CHECK CONSOLE
srv-01
SRV-01
```

````python
# D2
"""using .title(), .upper(), .lower(),to change a string case and proves those methods return a new string without changing the original"""
name = "record check console"
print(name.title()) #.tittle() capitalizes the first letter of each word
print(name.upper()) #changes texts to uppercase

original = "SRV-01"
print(original.lower())
print(original) #original did not change because we did not assign any string method so it remains itself
````

### D3. `.strip()`

1. Create `raw = "  srv-01  "`.
2. Print it inside square brackets so you can see the spaces: `print(f"[{raw}]")`
3. Print it again the same way, with `.strip()`.

**Expected output:** `[  srv-01  ]`, then `[srv-01]`

````python
# D3
"""using .strip() to remove extra spaces from the start and end of a string"""
raw = " srv-01 "
print(f"[{raw}]")   #no space is removed
print(f"[{raw.strip()}]")   #strip removes spaces
````

### D4. `.lstrip()`, `.rstrip()` and chaining

1. Create `messy = "***srv-01***"` and print `messy.strip("*")`, `messy.lstrip("*")` and `messy.rstrip("*")`.
2. Create `field = "  SRV-01  "` and print `field.strip().lower()` (two methods chained together).

**Expected output:** `srv-01`, `srv-01***`, `***srv-01`, then `srv-01`

````python
# D4
"""using .lstrip to remove spaces only on the left side and .rstrip only on the right side"""

messy = "***srv-01***"
print(messy.strip("*"))
print(messy.lstrip("*")) #removes spaces only on the left side
print(messy.rstrip("*")) #removes spaces only on the right side

field = " SRV-01"
print(field.strip().lower()) #this first removes spaces then changes it to lower case

````

## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
"""parsing one messy line, step by step"""

raw_line = " Srv-01 , used=87 , total=100 "
line = raw_line.strip() #removes leading and trailing whitespaces from the raw string
hostname_part, rest = line.split(",", 1)
hostname = hostname_part.strip().lower() #cleaning the hostname with .strip().lower()
data = {"hostname": hostname}
for piece in rest.split(","):
    key, value = piece.split("=")
    data[key.strip()] = float(value.strip())

print(data)
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15

def parse_line(raw_line): #defining the function that accepts a raw string argument
    """parse a messy raw server log line into a structured dictionary"""
    
    hostname_part, rest = raw_line.split(",", 1)
    hostname = hostname_part.strip().lower()

    data = {"hostname": hostname}
    for piece in rest.split(","):
        key, value = piece.split("=")
        data[key.strip()] = float(value.strip())
    return data #sends the final dictionary back to the caller

print(parse_line("SRV-02, used=120,total=100"))
````

*These combine several ideas from this week. There are no step-by-step instructions.*

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# C1
"""reading messy_lines.txt, reuse parse_line from D15 ,printing the hostname+status from used/total"""
with open("messy_lines.txt") as f:
    for raw_line in f:
        if not raw_line.strip():
            continue
        data = parse_line(raw_line)
        percent = data["used"] / data["total"] * 100

        if percent >= 100:
            status = "OVER LIMIT"
        elif percent >= 90:
            status = "WARNING"
        else:
            status = "OK"
        print(data["hostname"], status)
````

### C2. When parsing breaks

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]
```

**Expected output:**
```
{'hostname': 'srv-07', 'used': 50.0, 'total': 100.0}
Could not parse: srv-08,used=unknown,total=100
Could not parse: srv-09,note=a=b,used=1,total=2
```

````python
# C2
"""use try/except ValueError so bad lines print Could not parse: instead of crashing"""
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]

for line in lines:
    try:
        print(parse_line(line))
    except ValueError:
        print(f"Could not parse: {line}")
````
